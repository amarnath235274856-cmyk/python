```python
# get dummies into numerics
import pandas as pd
data={'names':['amarnath','sairam','hemu','anand',],
      'age':[19,20,21,22],
      'village':['enumulavaripalli','enumulavaripalli','enumulavaripalli','enumulavaripalli'],
      'work':['software','software','developer','developer']
}
result=pd.DataFrame(data,columns=['names','age','village','work'])
print(result)
```

```python
sample=pd.get_dummies(data['work'],dtype=int)
display(sample)
```
